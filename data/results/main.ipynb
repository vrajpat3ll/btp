```python
import pandas as pd

def summary(df):
    latency = df["latency_ms"]
    summary = pd.DataFrame({
        "Statistic": [
            "Mean",
            "Minimum",
            "25th percentile",
            "Median",
            "75th percentile",
            "90th percentile",
            "95th percentile",
            "Maximum",
        ],
        "Value": [
            latency.mean(),
            latency.min(),
            latency.quantile(0.25),
            latency.median(),
            latency.quantile(0.75),
            latency.quantile(0.90),
            latency.quantile(0.95),
            latency.max(),
        ]
    })

    # print(summary)
    return summary
```

```python
import os

BASE = "."   # your base directory

all_summaries = {}

for folder in os.listdir(BASE):
    path = os.path.join(BASE, folder)
    csv_path = os.path.join(path, "associative-latency.csv")

    if os.path.isdir(path) and os.path.exists(csv_path):
        df = pd.read_csv(csv_path)
        all_summaries[folder] = summary(df)
```

```python
for s, df in all_summaries.items():
    print(s)
    print('_'*20)
    print(df)
    print('*'*40)
```
